## Concept Notes
A base case stops recursion and prevents infinite calls.

Why this matters:
- Improves problem decomposition and state management.
- Builds confidence for interview-style search problems.
- Sharpens reasoning about stack behavior and pruning.


## Python Example
Run this snippet, then modify for a bigger input.


````python
def sum_to_n(n):
    if n == 0:  # base case
        return 0
    return n + sum_to_n(n - 1)

print(sum_to_n(5))
````

## Java Snippet
```java
static int factorial(int n) {
    if (n <= 1) return 1;
    return n * factorial(n - 1);
}
```
