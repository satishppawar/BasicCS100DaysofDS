## Basic Notes
What is Base Case?
- `Base Case` is a core DSA topic used to reason about correctness and efficiency.

Core theory details:
- Base case is the termination condition.
- Without it, recursion becomes infinite.
- Keep base case simple and easy to verify.

Why this matters for hands-on coding:

## Python Example
Run this snippet, then modify for a bigger input.


````python
def sum_to_n(n):
    if n == 0:  # base case
        return 0
    return n + sum_to_n(n - 1)

print(sum_to_n(5))
````

## Java Snippet

static int factorial(int n) {
