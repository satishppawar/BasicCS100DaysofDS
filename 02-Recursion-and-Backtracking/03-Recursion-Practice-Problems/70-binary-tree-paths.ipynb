## Basic Notes
What is Binary Tree Paths?
- `Binary Tree Paths` is a constraint-driven DSA problem typically solved with recursion/backtracking (or DP where applicable).
- A robust solution depends on state definition, decision choices, stopping rules, and pruning conditions.

Core theory details:
- Pattern: DFS root-to-leaf traversal.
- State: current node and path buffer.
- Complexity: O(n) traversal, plus path output cost.

## Python Example
Run this snippet, then modify for a bigger input.


````python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val, self.left, self.right = val, left, right


def paths(root):
    out = []

    def dfs(node, path):
        if not node:
            return
        path.append(str(node.val))
        if not node.left and not node.right:
            out.append("->".join(path))
        else:
            dfs(node.left, path)
            dfs(node.right, path)
        path.pop()

    dfs(root, [])
    return out

root = Node(1, Node(2, None, Node(5)), Node(3))
print(paths(root))
````

## Practice Solution Code (Java)
```java
import java.util.*;

class TreeNode {
    int val; TreeNode left, right;
    TreeNode(int v) { val = v; }
}

class BinaryTreePaths {
    static void dfs(TreeNode node, List<Integer> path, List<String> out) {
        if (node == null) return;
        path.add(node.val);
        if (node.left == null && node.right == null) {
            StringBuilder sb = new StringBuilder();
            for (int i = 0; i < path.size(); i++) {
                if (i > 0) sb.append("->");
                sb.append(path.get(i));
            }
            out.add(sb.toString());
        } else {
            dfs(node.left, path, out);
            dfs(node.right, path, out);
        }
        path.remove(path.size() - 1);
    }
}
```


````python
# Reference Solution: Binary Tree Paths

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def binary_tree_paths(root):
    if not root:
        return []

    out = []

    def dfs(node, path):
        if not node:
            return

        path.append(str(node.val))

        if not node.left and not node.right:
            out.append("->".join(path))
        else:
            dfs(node.left, path)
            dfs(node.right, path)

        path.pop()

    dfs(root, [])
    return out

root = TreeNode(1, TreeNode(2, None, TreeNode(5)), TreeNode(3))
print(binary_tree_paths(root))
````
