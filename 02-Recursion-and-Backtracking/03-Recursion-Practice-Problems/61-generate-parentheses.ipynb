## Python Example
Run this snippet, then modify for a bigger input.


````python
def generate_parenthesis(n):
    out, path = [], []

    def dfs(opened, closed):
        if len(path) == 2 * n:
            out.append("".join(path)); return
        if opened < n:
            path.append("("); dfs(opened + 1, closed); path.pop()
        if closed < opened:
            path.append(")"); dfs(opened, closed + 1); path.pop()

    dfs(0, 0)
    return out

print(generate_parenthesis(3))
````

````python
# Reference Solution: Generate Parentheses

def generate_parentheses(n):
    out = []
    path = []

    def dfs(opened, closed):
        if len(path) == 2 * n:
            out.append("".join(path))
            return
        if opened < n:
            path.append("(")
            dfs(opened + 1, closed)
            path.pop()
        if closed < opened:
            path.append(")")
            dfs(opened, closed + 1)
            path.pop()

    dfs(0, 0)
    return out

print(generate_parentheses(3))
````
