## Basic Notes
What is Combinations?
- `Combinations` is a constraint-driven DSA problem typically solved with recursion/backtracking (or DP where applicable).
- A robust solution depends on state definition, decision choices, stopping rules, and pruning conditions.

Core theory details:
- Pattern: include/explore with increasing start index.
- State: start pointer, current path length.
- Complexity: proportional to number of generated combinations.

## Python Example
Run this snippet, then modify for a bigger input.


````python
def combine(n, k):
    out, path = [], []

    def dfs(start):
        if len(path) == k:
            out.append(path.copy()); return
        for x in range(start, n + 1):
            path.append(x); dfs(x + 1); path.pop()

    dfs(1)
    return out

print(combine(4, 2))
````

## Practice Solution Code (Java)
```java
import java.util.*;

class Combinations {
    static void dfs(int n, int k, int start, List<Integer> path, List<List<Integer>> out) {
        if (path.size() == k) {
            out.add(new ArrayList<>(path));
            return;
        }
        for (int x = start; x <= n; x++) {
            path.add(x);
            dfs(n, k, x + 1, path, out);
            path.remove(path.size() - 1);
        }
    }
}
```


````python
# Reference Solution: Combinations

def combine(n, k):
    out = []
    path = []

    def dfs(start):
        if len(path) == k:
            out.append(path.copy())
            return
        # Prune when remaining numbers are insufficient.
        max_start = n - (k - len(path)) + 1
        for value in range(start, max_start + 1):
            path.append(value)
            dfs(value + 1)
            path.pop()

    dfs(1)
    return out

print(combine(4, 2))
````
