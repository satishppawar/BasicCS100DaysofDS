## Practice Solution Code (Java)
```java
class SudokuSolver {
    static boolean solve(char[][] board) {
        for (int r = 0; r < 9; r++) {
            for (int c = 0; c < 9; c++) {
                if (board[r][c] != '.') continue;
                for (char d = '1'; d <= '9'; d++) {
                    if (valid(board, r, c, d)) {
                        board[r][c] = d;
                        if (solve(board)) return true;
                        board[r][c] = '.';
                    }
                }
                return false;
            }
        }
        return true;
    }

    static boolean valid(char[][] b, int r, int c, char d) {
        for (int i = 0; i < 9; i++) {
            if (b[r][i] == d || b[i][c] == d) return false;
            int br = 3 * (r / 3) + i / 3;
            int bc = 3 * (c / 3) + i % 3;
            if (b[br][bc] == d) return false;
        }
        return true;
    }
}
```


````python
# Reference Solution: Sudoku Solver

def solve_sudoku(board):
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]
    empties = []

    for r in range(9):
        for c in range(9):
            value = board[r][c]
            if value == ".":
                empties.append((r, c))
            else:
                rows[r].add(value)
                cols[c].add(value)
                boxes[(r // 3) * 3 + (c // 3)].add(value)

    digits = "123456789"

    def dfs(index):
        if index == len(empties):
            return True

        r, c = empties[index]
        b = (r // 3) * 3 + (c // 3)

        for d in digits:
            if d in rows[r] or d in cols[c] or d in boxes[b]:
                continue

            board[r][c] = d
            rows[r].add(d)
            cols[c].add(d)
            boxes[b].add(d)

            if dfs(index + 1):
                return True

            board[r][c] = "."
            rows[r].remove(d)
            cols[c].remove(d)
            boxes[b].remove(d)

        return False

    dfs(0)
    return board

sample = [
    ["5", "3", ".", ".", "7", ".", ".", ".", "."],
    ["6", ".", ".", "1", "9", "5", ".", ".", "."],
    [".", "9", "8", ".", ".", ".", ".", "6", "."],
    ["8", ".", ".", ".", "6", ".", ".", ".", "3"],
    ["4", ".", ".", "8", ".", "3", ".", ".", "1"],
    ["7", ".", ".", ".", "2", ".", ".", ".", "6"],
    [".", "6", ".", ".", ".", ".", "2", "8", "."],
    [".", ".", ".", "4", "1", "9", ".", ".", "5"],
    [".", ".", ".", ".", "8", ".", ".", "7", "9"],
]

solved = solve_sudoku(sample)
for row in solved:
    print(" ".join(row))
````
