## Basic Notes
What is N-Queens Problem?
- `N-Queens Problem` is a constraint-driven DSA problem typically solved with recursion/backtracking (or DP where applicable).
- A robust solution depends on state definition, decision choices, stopping rules, and pruning conditions.

Core theory details:
- Pattern: row-wise backtracking with constraint pruning.
- State: used columns and diagonals.
- Complexity: exponential search; pruning is essential.

## Python Example
Run this snippet, then modify for a bigger input.


````python
def solve_n_queens(n):
    cols, diag1, diag2 = set(), set(), set()
    board = [["."] * n for _ in range(n)]
    out = []

    def dfs(r):
        if r == n:
            out.append(["".join(row) for row in board]); return
        for c in range(n):
            if c in cols or (r - c) in diag1 or (r + c) in diag2:
                continue
            cols.add(c); diag1.add(r - c); diag2.add(r + c)
            board[r][c] = "Q"
            dfs(r + 1)
            board[r][c] = "."
            cols.remove(c); diag1.remove(r - c); diag2.remove(r + c)

    dfs(0)
    return out

print(len(solve_n_queens(4)))
````

```java
import java.util.*;

class NQueens {
    static void dfs(int n, int row, Set<Integer> cols, Set<Integer> d1, Set<Integer> d2, char[][] board, List<List<String>> out) {
        if (row == n) {
            List<String> sol = new ArrayList<>();
            for (char[] r : board) sol.add(new String(r));
            out.add(sol);
            return;
        }
        for (int c = 0; c < n; c++) {
            if (cols.contains(c) || d1.contains(row - c) || d2.contains(row + c)) continue;
            cols.add(c); d1.add(row - c); d2.add(row + c);
            board[row][c] = 'Q';
            dfs(n, row + 1, cols, d1, d2, board, out);
            board[row][c] = '.';
            cols.remove(c); d1.remove(row - c); d2.remove(row + c);
        }
    }
}
```


````python
# Reference Solution: N-Queens

def solve_n_queens(n):
    board = [["."] * n for _ in range(n)]
    used_cols = set()
    used_diag1 = set()  # row - col
    used_diag2 = set()  # row + col
    out = []

    def dfs(row):
        if row == n:
            out.append(["".join(r) for r in board])
            return

        for col in range(n):
            if col in used_cols or (row - col) in used_diag1 or (row + col) in used_diag2:
                continue

            used_cols.add(col)
            used_diag1.add(row - col)
            used_diag2.add(row + col)
            board[row][col] = "Q"

            dfs(row + 1)

            board[row][col] = "."
            used_cols.remove(col)
            used_diag1.remove(row - col)
            used_diag2.remove(row + col)

    dfs(0)
    return out

solutions = solve_n_queens(4)
print("Count:", len(solutions))
print(solutions[0])
````
