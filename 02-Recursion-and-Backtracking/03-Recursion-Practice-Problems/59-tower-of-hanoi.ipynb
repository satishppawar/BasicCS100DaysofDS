## Basic Notes
What is Tower of Hanoi?
- `Tower of Hanoi` is a constraint-driven DSA problem typically solved with recursion/backtracking (or DP where applicable).
- A robust solution depends on state definition, decision choices, stopping rules, and pruning conditions.

Core theory details:
- Pattern: classic recursion with 2 subproblems + 1 move.
- State: number of disks and 3 rods.
- Complexity: O(2^n) time, O(n) recursion depth.

Why this matters for hands-on coding:

## Python Example
Run this snippet, then modify for a bigger input.


````python
def hanoi(n, src, dst, aux, moves):
    if n == 1:
        moves.append((src, dst)); return
    hanoi(n - 1, src, aux, dst, moves)
    moves.append((src, dst))
    hanoi(n - 1, aux, dst, src, moves)

moves = []
hanoi(3, 'A', 'C', 'B', moves)
print(moves)
````

## Practice Solution Code (Java)
```java
import java.util.*;

class TowerOfHanoi {
    static void solve(int n, char src, char dst, char aux, List<String> moves) {
        if (n == 1) {
            moves.add(src + "->" + dst);
            return;
        }
        solve(n - 1, src, aux, dst, moves);
        moves.add(src + "->" + dst);
        solve(n - 1, aux, dst, src, moves);
    }

    public static void main(String[] args) {
        List<String> moves = new ArrayList<>();
        solve(3, 'A', 'C', 'B', moves);
        System.out.println(moves);
    }
}
```


````python
# Reference Solution: Tower of Hanoi

def tower_of_hanoi(n, source, target, auxiliary):
    moves = []

    def solve(k, src, dst, aux):
        if k == 1:
            moves.append((src, dst))
            return
        solve(k - 1, src, aux, dst)
        moves.append((src, dst))
        solve(k - 1, aux, dst, src)

    solve(n, source, target, auxiliary)
    return moves

result = tower_of_hanoi(3, "A", "C", "B")
print("Total moves:", len(result))
print(result)
````
