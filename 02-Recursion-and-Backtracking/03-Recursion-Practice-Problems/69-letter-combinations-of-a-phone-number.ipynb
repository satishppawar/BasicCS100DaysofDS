````python
def letter_combinations(digits):
    if not digits:
        return []
    mp = {"2":"abc","3":"def","4":"ghi","5":"jkl","6":"mno","7":"pqrs","8":"tuv","9":"wxyz"}
    out, path = [], []

    def dfs(i):
        if i == len(digits):
            out.append("".join(path)); return
        for ch in mp[digits[i]]:
            path.append(ch); dfs(i + 1); path.pop()

    dfs(0)
    return out

print(letter_combinations("23"))
````

## Practice Solution Code (Java)
```java
import java.util.*;

class PhoneCombinations {
    static final String[] MAP = {"", "", "abc", "def", "ghi", "jkl", "mno", "pqrs", "tuv", "wxyz"};

    static void dfs(String digits, int i, StringBuilder path, List<String> out) {
        if (i == digits.length()) {
            out.add(path.toString());
            return;
        }
        String letters = MAP[digits.charAt(i) - '0'];
        for (char ch : letters.toCharArray()) {
            path.append(ch);
            dfs(digits, i + 1, path, out);
            path.deleteCharAt(path.length() - 1);
        }
    }
}
```


````python
# Reference Solution: Letter Combinations of a Phone Number

def letter_combinations(digits):
    if not digits:
        return []

    keypad = {
        "2": "abc",
        "3": "def",
        "4": "ghi",
        "5": "jkl",
        "6": "mno",
        "7": "pqrs",
        "8": "tuv",
        "9": "wxyz",
    }

    out = []
    path = []

    def dfs(index):
        if index == len(digits):
            out.append("".join(path))
            return

        for ch in keypad[digits[index]]:
            path.append(ch)
            dfs(index + 1)
            path.pop()

    dfs(0)
    return out

print(letter_combinations("23"))
````
