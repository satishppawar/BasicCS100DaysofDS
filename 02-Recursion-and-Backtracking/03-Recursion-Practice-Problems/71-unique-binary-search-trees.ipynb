## Basic Notes
What is Unique Binary Search Trees?
- `Unique Binary Search Trees` is a constraint-driven DSA problem typically solved with recursion/backtracking (or DP where applicable).
- A robust solution depends on state definition, decision choices, stopping rules, and pruning conditions.

Core theory details:
- Pattern: Catalan recurrence / dynamic programming.
- State: number of nodes as subproblem.
- Complexity: O(n^2) time, O(n) space with DP.

Why this matters for hands-on coding:

## Python Example
Run this snippet, then modify for a bigger input.


````python
def num_trees(n):
    dp = [0] * (n + 1)
    dp[0] = dp[1] = 1
    for nodes in range(2, n + 1):
        for root in range(1, nodes + 1):
            dp[nodes] += dp[root - 1] * dp[nodes - root]
    return dp[n]

print(num_trees(3))
````

```java
class UniqueBSTCount {
    static int numTrees(int n) {
        int[] dp = new int[n + 1];
        dp[0] = 1;
        dp[1] = 1;
        for (int nodes = 2; nodes <= n; nodes++) {
            int total = 0;
            for (int root = 1; root <= nodes; root++) {
                total += dp[root - 1] * dp[nodes - root];
            }
            dp[nodes] = total;
        }
        return dp[n];
    }
}
```


````python
# Reference Solution: Unique Binary Search Trees (count)

def num_trees(n):
    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1

    for nodes in range(2, n + 1):
        total = 0
        for root in range(1, nodes + 1):
            left_count = dp[root - 1]
            right_count = dp[nodes - root]
            total += left_count * right_count
        dp[nodes] = total

    return dp[n]

print(num_trees(3))
````
