## Basic Notes
What is Combination Sum III?
- `Combination Sum III` is a constraint-driven DSA problem typically solved with recursion/backtracking (or DP where applicable).
- A robust solution depends on state definition, decision choices, stopping rules, and pruning conditions.

Core theory details:
- Pattern: choose k distinct numbers from 1..9.
- State: next number, remaining sum, picks count.
- Complexity: bounded exponential due to small search space.

## Python Example
Run this snippet, then modify for a bigger input.


````python
def combination_sum3(k, n):
    out, path = [], []

    def dfs(start, remain):
        if len(path) == k and remain == 0:
            out.append(path.copy()); return
        if len(path) >= k or remain < 0:
            return
        for x in range(start, 10):
            path.append(x)
            dfs(x + 1, remain - x)
            path.pop()

    dfs(1, n)
    return out

print(combination_sum3(3, 7))
````

## Practice Solution Code (Java)
```java
import java.util.*;

class CombinationSumIII {
    static void dfs(int start, int k, int remain, List<Integer> path, List<List<Integer>> out) {
        if (path.size() == k) {
            if (remain == 0) out.add(new ArrayList<>(path));
            return;
        }
        for (int x = start; x <= 9; x++) {
            if (x > remain) break;
            path.add(x);
            dfs(x + 1, k, remain - x, path, out);
            path.remove(path.size() - 1);
        }
    }
}
```


````python
# Reference Solution: Combination Sum III

def combination_sum_iii(k, n):
    out = []
    path = []

    def dfs(start, remain):
        if len(path) == k:
            if remain == 0:
                out.append(path.copy())
            return

        for value in range(start, 10):
            if value > remain:
                break
            path.append(value)
            dfs(value + 1, remain - value)
            path.pop()

    dfs(1, n)
    return out

print(combination_sum_iii(3, 9))
````
