Core theory details:

## Python Example
Run this snippet, then modify for a bigger input.


````python
# Python does not optimize tail recursion automatically.
# Convert tail recursion to loop when needed.

def factorial_loop(n):
    acc = 1
    while n > 1:
        acc *= n
        n -= 1
    return acc

print(factorial_loop(6))
````

class FactorialDemo {
    static int factorial(int n) {
        if (n <= 1) return 1;
        return n * factorial(n - 1);
    }

    static int factorialTail(int n, int acc) {
        if (n <= 1) return acc;
        return factorialTail(n - 1, acc * n);
    }
}
