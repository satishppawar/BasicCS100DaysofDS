## Python Example
Run this snippet, then modify for a bigger input.


````python
def is_even(n):
    if n == 0:
        return True
    return is_odd(n - 1)


def is_odd(n):
    if n == 0:
        return False
    return is_even(n - 1)

print(is_even(8), is_odd(9))
````
