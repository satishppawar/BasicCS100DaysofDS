## Python Example
Run this snippet, then modify for a bigger input.


````python
def tree_recursion(n):
    if n == 0:
        return 1
    left = tree_recursion(n - 1)
    right = tree_recursion(n - 1)
    return left + right

print(tree_recursion(3))
````
