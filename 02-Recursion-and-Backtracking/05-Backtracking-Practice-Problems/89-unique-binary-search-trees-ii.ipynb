## Python Example
Run this snippet, then modify for a bigger input.


````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val, self.left, self.right = val, left, right


def generate(start, end):
    if start > end:
        return [None]
    trees = []
    for root in range(start, end + 1):
        for left in generate(start, root - 1):
            for right in generate(root + 1, end):
                trees.append(TreeNode(root, left, right))
    return trees

print(len(generate(1, 3)))
````

## Practice Solution Code (Java)

```java
import java.util.*;

class TreeNode {
    int val; TreeNode left, right;
    TreeNode(int v) { val = v; }
    TreeNode(int v, TreeNode l, TreeNode r) { val = v; left = l; right = r; }
}

class UniqueBSTStructures {
    static List<TreeNode> build(int start, int end) {
        List<TreeNode> out = new ArrayList<>();
        if (start > end) {
            out.add(null);
            return out;
        }
        for (int root = start; root <= end; root++) {
            for (TreeNode l : build(start, root - 1)) {
                for (TreeNode r : build(root + 1, end)) {
                    out.add(new TreeNode(root, l, r));
                }
            }
        }
        return out;
    }
}
```


````python
# Reference Solution: Unique Binary Search Trees II (generate trees)

from functools import lru_cache

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


@lru_cache(maxsize=None)
def build(start, end):
    if start > end:
        return (None,)

    trees = []
    for root_value in range(start, end + 1):
        left_trees = build(start, root_value - 1)
        right_trees = build(root_value + 1, end)

        for left in left_trees:
            for right in right_trees:
                trees.append(TreeNode(root_value, left, right))

    return tuple(trees)


def generate_trees(n):
    if n == 0:
        return []
    return list(build(1, n))


def serialize_preorder(root):
    if not root:
        return "null"
    return f"{root.val},{serialize_preorder(root.left)},{serialize_preorder(root.right)}"

all_trees = generate_trees(3)
print("Count:", len(all_trees))
print("First tree:", serialize_preorder(all_trees[0]))
````
