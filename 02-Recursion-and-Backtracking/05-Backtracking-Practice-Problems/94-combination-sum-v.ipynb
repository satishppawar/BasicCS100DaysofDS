## Python Example
Run this snippet, then modify for a bigger input.


````python
# Design your own variant:
# Example rule: fixed length combinations with repetition allowed.

def combo_sum_v(nums, target, length):
    out, path = [], []

    def dfs(start, remain):
        if len(path) == length:
            if remain == 0:
                out.append(path.copy())
            return
        for i in range(start, len(nums)):
            if remain - nums[i] < 0:
                continue
            path.append(nums[i])
            dfs(i, remain - nums[i])
            path.pop()

    dfs(0, target)
    return out

print(combo_sum_v([1,2,3], 4, 2))
````

## Practice Solution Code (Java)
```java
import java.util.*;

class CombinationSumV {
    // Custom variant: exactly k picks, repetition allowed, order ignored.
    static void dfs(int[] nums, int start, int remain, int picksLeft, List<Integer> path, List<List<Integer>> out) {
        if (picksLeft == 0) {
            if (remain == 0) out.add(new ArrayList<>(path));
            return;
        }
        for (int i = start; i < nums.length; i++) {
            if (nums[i] > remain) break;
            path.add(nums[i]);
            dfs(nums, i, remain - nums[i], picksLeft - 1, path, out);
            path.remove(path.size() - 1);
        }
    }
}
```


````python
# Reference Solution: Combination Sum V (custom variant)
# Variant definition:
# Find combinations of EXACTLY k numbers (repetition allowed) that sum to target.
# Order does not matter.

def combination_sum_v(nums, target, k):
    nums = sorted(nums)
    out = []
    path = []

    def dfs(start, remain, picks_left):
        if picks_left == 0:
            if remain == 0:
                out.append(path.copy())
            return

        for i in range(start, len(nums)):
            value = nums[i]
            if value > remain:
                break
            path.append(value)
            dfs(i, remain - value, picks_left - 1)
            path.pop()

    dfs(0, target, k)
    return out

print(combination_sum_v([1, 2, 3], 6, 3))
````
