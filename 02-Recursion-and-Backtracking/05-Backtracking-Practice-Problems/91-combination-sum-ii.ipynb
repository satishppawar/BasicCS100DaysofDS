## Python Example
Run this snippet, then modify for a bigger input.


````python
def combination_sum2(candidates, target):
    candidates.sort()
    out, path = [], []

    def dfs(start, remain):
        if remain == 0:
            out.append(path.copy()); return
        if remain < 0:
            return
        prev = None
        for i in range(start, len(candidates)):
            if candidates[i] == prev:
                continue
            path.append(candidates[i])
            dfs(i + 1, remain - candidates[i])
            path.pop()
            prev = candidates[i]

    dfs(0, target)
    return out

print(combination_sum2([10,1,2,7,6,1,5], 8))
````

## Practice Solution Code (Java)
```java
import java.util.*;

class CombinationSumII {
    static void dfs(int[] a, int start, int remain, List<Integer> path, List<List<Integer>> out) {
        if (remain == 0) {
            out.add(new ArrayList<>(path));
            return;
        }
        for (int i = start; i < a.length; i++) {
            if (i > start && a[i] == a[i - 1]) continue;
            if (a[i] > remain) break;
            path.add(a[i]);
            dfs(a, i + 1, remain - a[i], path, out);
            path.remove(path.size() - 1);
        }
    }
}
```


````python
# Reference Solution: Combination Sum II

def combination_sum_ii(candidates, target):
    candidates.sort()
    out = []
    path = []

    def dfs(start, remain):
        if remain == 0:
            out.append(path.copy())
            return

        prev = None
        for i in range(start, len(candidates)):
            value = candidates[i]
            if value == prev:
                continue
            if value > remain:
                break

            path.append(value)
            dfs(i + 1, remain - value)
            path.pop()
            prev = value

    dfs(0, target)
    return out

print(combination_sum_ii([10, 1, 2, 7, 6, 1, 5], 8))
````
