## Basic Notes
What is Palindrome Partitioning?
- `Palindrome Partitioning` is a constraint-driven DSA problem typically solved with recursion/backtracking (or DP where applicable).
- A robust solution depends on state definition, decision choices, stopping rules, and pruning conditions.

Core theory details:
- Pattern: backtracking over cut positions.
- State: start index and current partitions.
- Complexity: exponential; palindrome checks can be optimized.

## Python Example
Run this snippet, then modify for a bigger input.


````python
def partition(s):
    out, path = [], []

    def is_pal(x):
        return x == x[::-1]

    def dfs(start):
        if start == len(s):
            out.append(path.copy()); return
        for end in range(start + 1, len(s) + 1):
            part = s[start:end]
            if is_pal(part):
                path.append(part); dfs(end); path.pop()

    dfs(0)
    return out

print(partition("aab"))
````

## Practice Solution Code (Java)
```java
import java.util.*;

class PalindromePartitioning {
    static void dfs(String s, int start, List<String> path, List<List<String>> out) {
        if (start == s.length()) {
            out.add(new ArrayList<>(path));
            return;
        }
        for (int end = start; end < s.length(); end++) {
            if (!isPal(s, start, end)) continue;
            path.add(s.substring(start, end + 1));
            dfs(s, end + 1, path, out);
            path.remove(path.size() - 1);
        }
    }

    static boolean isPal(String s, int l, int r) {
        while (l < r) if (s.charAt(l++) != s.charAt(r--)) return false;
        return true;
    }
}
```


````python
# Reference Solution: Palindrome Partitioning

def partition(s):
    out = []
    path = []

    def is_palindrome(left, right):
        while left < right:
            if s[left] != s[right]:
                return False
            left += 1
            right -= 1
        return True

    def dfs(start):
        if start == len(s):
            out.append(path.copy())
            return

        for end in range(start, len(s)):
            if is_palindrome(start, end):
                path.append(s[start : end + 1])
                dfs(end + 1)
                path.pop()

    dfs(0)
    return out

print(partition("aab"))
````
