Why this matters:
- Improves problem decomposition and state management.

- Sharpens reasoning about stack behavior and pruning.


## Python Example
Run this snippet, then modify for a bigger input.


````python
def can_make_sum(nums, target):
    def dfs(i, total):
        if total == target:
            return True
        if i == len(nums) or total > target:
            return False
        return dfs(i + 1, total + nums[i]) or dfs(i + 1, total)

    return dfs(0, 0)

print(can_make_sum([2, 3, 7], 10))
````
