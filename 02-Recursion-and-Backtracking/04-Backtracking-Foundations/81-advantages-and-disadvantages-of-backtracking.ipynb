## Basic Notes
What is Advantages and Disadvantages of Backtracking?
- `Advantages and Disadvantages of Backtracking` focuses on exploring candidate solutions and undoing choices when constraints fail.
- Performance depends on pruning effectiveness and the quality of state representation.

Core theory details:
- Pros: complete search with clear structure.
- Cons: exponential worst-case behavior.
- Pruning heuristics decide practical performance.

## Python Example
Run this snippet, then modify for a bigger input.


````python
def can_make_sum(nums, target):
    def dfs(i, total):
        if total == target:
            return True
        if i == len(nums) or total > target:
            return False
        return dfs(i + 1, total + nums[i]) or dfs(i + 1, total)

    return dfs(0, 0)

print(can_make_sum([2, 3, 7], 10))
````

import java.util.*;

class BacktrackingTemplate {
    static void dfs(int[] nums, int i, List<Integer> path, List<List<Integer>> out) {
        if (i == nums.length) {
            out.add(new ArrayList<>(path));
            return;
        }
        path.add(nums[i]);
        dfs(nums, i + 1, path, out);
        path.remove(path.size() - 1);
        dfs(nums, i + 1, path, out);
    }
}
