## Python Example
Run this snippet, then modify for a bigger input.


````python
def generate_binary(n):
    out = []
    path = []

    def dfs(i):
        if i == n:
            out.append("".join(path)); return
        for ch in "01":
            path.append(ch); dfs(i + 1); path.pop()

    dfs(0)
    return out

print(generate_binary(3))
````

## Practice Solution Code (Java)

```java
import java.util.*;

class BacktrackingTemplate {
    static void dfs(int[] nums, int i, List<Integer> path, List<List<Integer>> out) {
        if (i == nums.length) {
            out.add(new ArrayList<>(path));
            return;
        }
        path.add(nums[i]);
        dfs(nums, i + 1, path, out);
        path.remove(path.size() - 1);
        dfs(nums, i + 1, path, out);
    }
}
```
