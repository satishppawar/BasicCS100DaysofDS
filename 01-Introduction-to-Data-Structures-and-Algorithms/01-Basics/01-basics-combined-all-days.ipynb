# Day 02 - Variables: System Defined vs User Defined

## Learning Objectives
- Understand the core idea behind **Variables: System Defined vs User Defined**.
- Connect the concept to coding interviews and production engineering decisions.
- Apply the concept with one runnable example and one practice implementation.

## Introduction
In programming, variables are fundamental building blocks that allow us to store and manipulate data. There are two main types of variables: system-defined and user-defined. Understanding the differences between these types of variables is crucial for effective programming.

- **System Defined Variables**: These are variables that are predefined by the programming language or environment. They often have special meanings and are used for specific purposes. Examples include `null`, `undefined`, `NaN` in JavaScript, or `None` in Python.

- **User Defined Variables**: These are variables that are created by the programmer to store data. They can be named according to the programmer's choice and can hold various types of data, such as integers, strings, or objects.

- **Key Differences**:
    - **Definition**: System-defined variables are built into the programming language, while user-defined variables are created by the programmer.
    - **Purpose**: System-defined variables often serve specific functions within the language, while user-defined variables are used to store and manipulate data as needed by the programmer.
    - **Naming**: System-defined variables have fixed names, while user-defined variables can be named according to the programmer's preferences, following the language's naming conventions.

## Basic Notes
What is Variables: System Defined vs User Defined?
- System-defined types are built-in primitives and collections provided by the language runtime.
- User-defined types are custom classes/structures that model domain-specific behavior and data.

Core theory details:
- System-defined types are built-in primitives/collections.
- Examples: int, float, str, list, dict, set.
- User-defined types model domain objects and behavior.
- Choosing the right type early reduces conversions, validation bugs, and code complexity.

Why this matters for hands-on coding:
- Use system-defined types for low-level values and utility containers.
- Use user-defined types when business rules and invariants must live with the data model.

```python
# Day 02 - Python Example: System Defined vs User Defined Variables
import sys

system_defined = 42
print("System Defined variable: type of variable is",  type(system_defined).__name__, "Size is", sys.getsizeof(system_defined), "and value is", system_defined, "\n")

class UserDefined:
    def __init__(self, name):
        self.name = name

u = UserDefined("node")
print("User Defined variable: type of variable is", type(u).__name__, "Size is", sys.getsizeof(u), "and value is", u.name)
```

```python
# Day 06 - Python Example: Algorithms (Linear Search)
def linear_search(nums, target):
    for i, value in enumerate(nums):
        if value == target:
            return i
    return -1

print(linear_search([4, 8, 15, 16, 23, 42], 23))
```
