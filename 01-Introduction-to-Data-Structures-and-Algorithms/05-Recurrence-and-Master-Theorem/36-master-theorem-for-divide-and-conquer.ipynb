## Python Example
Run and modify this snippet.


````python
def classify_master(a, b, k):
    import math
    critical = math.log(a, b)
    if k < critical:
        return "Case 1: Theta(n^(log_b(a)))"
    if k == critical:
        return "Case 2: Theta(n^k log n)"
    return "Case 3: Theta(n^k)"

print(classify_master(2, 2, 1))
````
