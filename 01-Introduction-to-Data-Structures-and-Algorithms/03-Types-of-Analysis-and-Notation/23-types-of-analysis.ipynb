## Python Example
Run and modify this snippet.


````python
def cost_linear_search(position, n):
    if position < 0:
        return n
    return position + 1

n = 10
print("best", cost_linear_search(0, n))
print("worst", cost_linear_search(-1, n))
print("average approx", n / 2)
````
