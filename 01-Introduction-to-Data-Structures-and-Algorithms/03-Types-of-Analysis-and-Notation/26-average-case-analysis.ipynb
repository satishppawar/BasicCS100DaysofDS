## Python Example
Run and modify this snippet.


````python
import random

def average_hits(trials=1000, n=100):
    total_steps = 0
    for _ in range(trials):
        target = random.randint(0, n - 1)
        total_steps += target + 1
    return total_steps / trials

print(round(average_hits(), 2))
````
