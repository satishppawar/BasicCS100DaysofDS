## Python Example
Run and verify this example.


````python
class Node:
    def __init__(self, v, l=None, r=None):
        self.v = v; self.l = l; self.r = r

root = Node(2, Node(1), Node(3))
def inorder(n):
    return inorder(n.l) + [n.v] + inorder(n.r) if n else []
print(inorder(root))
````
