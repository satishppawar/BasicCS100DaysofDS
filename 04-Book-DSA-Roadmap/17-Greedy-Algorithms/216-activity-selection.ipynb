## Python Example

````python
intervals = [(1,3),(2,4),(3,5),(0,6),(5,7),(8,9)]
intervals.sort(key=lambda x: x[1])
count = 0
end = -10**9
for s,e in intervals:
    if s >= end:
        count += 1
        end = e
print(count)
````
