## Python Example
Run and verify this example.


````python
parent = list(range(6))
rank = [0]*6
def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]
def union(a,b):
    ra, rb = find(a), find(b)
    if ra == rb: return
    if rank[ra] < rank[rb]: parent[ra] = rb
    elif rank[ra] > rank[rb]: parent[rb] = ra
    else: parent[rb] = ra; rank[ra] += 1
union(1,2); union(2,3)
print(find(1) == find(3))
````
