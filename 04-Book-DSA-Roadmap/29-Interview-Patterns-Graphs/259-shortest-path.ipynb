## Python Example
Run and verify this example.


````python
import heapq
g = {0:[(1,4),(2,1)], 1:[(3,1)], 2:[(1,2),(3,5)], 3:[]}
dist = {0:0}
pq = [(0,0)]
while pq:
    d,u = heapq.heappop(pq)
    if d != dist.get(u, float('inf')): continue
    for v,w in g[u]:
        nd = d + w
        if nd < dist.get(v, float('inf')):
            dist[v] = nd
            heapq.heappush(pq, (nd, v))
print(dist)
````
