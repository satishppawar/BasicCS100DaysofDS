## Python Example
Run and verify this example.


````python
class Node:
    def __init__(self, val, nxt=None):
        self.val = val
        self.next = nxt

head = Node(1, Node(2, Node(3)))
cur = head
while cur:
    print(cur.val, end=" ")
    cur = cur.next
print()
````
